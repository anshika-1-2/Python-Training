```python
#cube example
import pandas as pd
import numpy as np
import timeit


df = pd.DataFrame({
    "values": np.random.randint(1, 100, size=1000)
})

def cube_loop(df):
    result = []
    for v in df["values"]:
        result.append(v ** 3)
    return result



def cube_apply(df):
    return df["values"].apply(lambda x: x ** 3)



def cube_vectorized(df):
    return df["values"] ** 3


loop_time = timeit.timeit(
    stmt="cube_loop(df)",
    globals=globals(),
    number=3
)

apply_time = timeit.timeit(
    stmt="cube_apply(df)",
    globals=globals(),
    number=3
)

vectorized_time = timeit.timeit(
    stmt="cube_vectorized(df)",
    globals=globals(),
    number=3
)




print("Execution Time (3 runs):")
print(f"Python loop       : {loop_time:.4f} seconds")
print(f"apply()           : {apply_time:.4f} seconds")
print(f"Vectorized Pandas : {vectorized_time:.4f} seconds")



df["cubed"] = cube_vectorized(df)
print("\nSample output:")
print(df.head())
```
